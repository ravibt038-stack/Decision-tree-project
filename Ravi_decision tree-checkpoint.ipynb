### Importing necessary libraries

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

### Data collection

```python
claimant_data=pd.read_csv("Claimaints.csv.csv")
claimant_data
```

### Data understanding

```python
claimant_data.isnull().sum()
```

### Data preparation

```python
claimant_data.dropna(inplace=True) #removing null values
```

```python
claimant_data.isnull().sum()
```

### Data preparation (seperating variables and logistic regression)

```python
y=claimant_data["ATTORNEY"]
x=claimant_data.iloc[:,2:]
```

```python
X_train, X_test, y_train, y_test = train_test_split(x,y,train_size=0.80,shuffle=True)
```

### Model bulding

```python
logistic_model=LogisticRegression()
```

### Model training

```python
logistic_model.fit(X_train,y_train) # this is for logistic regression not for decision tree
```

### Decision tree 

```python
from sklearn.tree import DecisionTreeClassifier 
```

```python
dt_model=DecisionTreeClassifier() # this model is decision tree model
```

```python
dt_model.fit(X_train,y_train)
```

### Model testing

#### Training data

```python
y_predict_train=dt_model.predict(X_train)
```

#### Testing data

```python
y_predict_test=dt_model.predict(X_test)
```

### Model evaluation

#### Train data evaluation

```python
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix, roc_auc_score, roc_curve
```

```python
print(classification_report(y_true=y_train, y_pred=y_predict_train))
```

```python
fpr, tpr, thresholds = roc_curve(y_train, dt_model.predict_proba (X_train) [:,1])

auc = roc_auc_score(y_train,y_predict_train)
print(auc)

import matplotlib.pyplot as plt
plt.plot(fpr, tpr, color='red', label='logit model ( area = %0.2f) '%auc)
plt.legend()
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate or [1 - True Negative Rate]')
plt.ylabel('True Positive Rate')
```

#### Test data evaluation

```python
print(classification_report(y_true=y_test, y_pred=y_predict_test))
```

```python
accuracy_score(y_true=y_test, y_pred=y_predict_test)
```

```python
confusion_matrix(y_true=y_test, y_pred=y_predict_test)
```

```python
fpr, tpr, thresholds = roc_curve(y_test, dt_model.predict_proba (X_test) [:,1])

auc = roc_auc_score(y_test,y_predict_test)
print(auc)

import matplotlib.pyplot as plt
plt.plot(fpr, tpr, color='red', label='logit model ( area = %0.2f) '%auc)
plt.legend()
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate or [1 - True Negative Rate]')
plt.ylabel('True Positive Rate')
```
